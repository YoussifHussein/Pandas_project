```python
apps_df['released'] = pd.to_datetime(apps_df['released'], errors='coerce')
print(apps_df.dtypes)
```

```python
#10. Find the earliest and latest release date in the dataset.
released_date = apps_df.sort_values(by='released', ascending=False)
far_date = released_date['released'].iloc[1]
near_date = released_date['released'].iloc[-1]
print(f"The near released date is {near_date},and The far released date is {far_date}")
```

```python
#11. Calculate the average number of ratings per day for apps with more than 10,000 reviews.
apps_df["reviews_per_day"]= apps_df["reviews"]/365
average_number_reviews=apps_df.groupby("title")["reviews_per_day"].sum().sort_values(ascending=False)
print(average_number_reviews[average_number_reviews>10000])
```

```python
#12. Identify the top 3 genres with the highest average app score.
apps_df["ave_score"]=apps_df["score"].mean()
apps_df.groupby("genre")["ave_score"].sum().sort_values(ascending=False).head(3)
```

```python
#13. Determine the percentage of apps that are ad-supported and offer in-app purchases 
total_apps = len(apps_df)
count_both = ((apps_df["offers_iap"] == True) & (apps_df["ad_supported"] == True)).sum()
percentage = (count_both / total_apps) * 100
print("percentage is:",percentage,"%") 
```

```python
#14. Find the most common release year for apps in the dataset.
most_common_year = apps_df['released'].dt.year.value_counts().idxmax()
most_common_year
```

```python
#15. Calculate the correlation between the number of reviews and the average rating for each genre.
correlation = apps_df.groupby("genre")[["reviews", "score"]].corr()
print(correlation)
```

```python
#16. Calculate the weighted average rating for each genre, where the weight is the number of reviews. 
def weighted_avg(group):
    return (group["score"] * group["reviews"]).sum() / group["reviews"].sum()
weighted = apps_df.groupby("genre").apply(
    weighted_avg, include_groups=False
).reset_index()
weighted.columns = ["genre", "weighted_avg_rating"]
weighted["weighted_avg_rating"] = weighted["weighted_avg_rating"].round(2)
weighted = weighted.sort_values("weighted_avg_rating", ascending=False)
print(weighted)
```

```python
#17. Identify the app with the highest ratio of ratings per review for each genr
apps_df['rating_review_ratio'] = apps_df['ratings'] / apps_df['reviews']

best_ratio_apps = apps_df.loc[
    apps_df.groupby('genre')['rating_review_ratio'].idxmax()
][['genre', 'title', 'rating_review_ratio']]

best_ratio_apps.head()
```
